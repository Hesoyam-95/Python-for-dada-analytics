```python
a = 17
b = 5
print (a + b)
print (a // b)
print (a % b)

# Додавання
# Ціла частина від ділення
# Відсоток від ділення
```

"+"	додавання	17 + 5 = 22


"-"	віднімання	17 - 5 = 12


"*"	множення	17 * 5 = 85


"/"	 ділення	17 / 5 = 3.4


"//"	ціла частина ділення	17 // 5 = 3


"%"	залишок	17 % 5 = 2
